```python
import networkx as nx
from networkx.algorithms import bipartite
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
#taking logoff dataset as graph parameters. No null values in this dataset
print(logoff); logoff.shape
```

```python
users_nodes = list(logoff['user'].unique())
pc_nodes = list(logoff['pc'].unique())
weighted_edges = [(row['user'], row['pc'], row['total']) for idx, row in logoff.iterrows()]
#weighted_edges
type(weighted_edges)
len(weighted_edges)
```

```python
graph = nx.Graph()
graph.add_nodes_from(users_nodes, bipartite = 0) # users as nodes
graph.add_nodes_from(pc_nodes, bipartite = 1) # PCs as nodes
graph.add_weighted_edges_from(weighted_edges)
```

```python
nx.is_connected(graph) #this shows that the graph isn't fully connected
```

```python
nx.is_bipartite(graph) # this shows that the graph is bipartite
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.svm import OneClassSVM

# Assuming user_pc is already defined
user_pc_count = np.array(user_pc['pc_count'])

# Reshape the data
user_pc_ct = user_pc_count.reshape(-1, 1)

# Define the One-Class SVM model
oc_svm = OneClassSVM(kernel='rbf', gamma='auto', nu=0.1)

# Fit the model
oc_svm.fit(user_pc_ct)

# Get anomaly scores (decision function values)
graph_a_score = oc_svm.decision_function(user_pc_ct)

# Predict anomalies (1 for normal, -1 for anomaly)
predictions = oc_svm.predict(user_pc_ct)

# Create a DataFrame for results
graph_result = pd.DataFrame()
graph_result['user'] = user_pc['user']
graph_result['pc_count'] = user_pc['pc_count']
graph_result['ascore'] = graph_a_score
graph_result['anomaly'] = predictions

# Identify outliers
outliers = graph_result.loc[graph_result['anomaly'] == -1]

# Plotting the data
plt.figure(figsize=(10, 6))

# Plot the normal data points
sns.scatterplot(x=graph_result.index, y=graph_result['pc_count'], hue=graph_result['anomaly'],
                palette={1: 'blue', -1: 'red'}, s=100)

# Highlight the outliers
plt.scatter(outliers.index, outliers['pc_count'], color='red', label='Anomalies', edgecolor='black', s=150)

# Adding labels and title
plt.title('One-Class SVM Anomaly Detection', fontsize=16)
plt.xlabel('User Index', fontsize=14)
plt.ylabel('PC Count', fontsize=14)

# Show legend
plt.legend(['Normal', 'Anomaly'], loc='upper left')

# Display the plot
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd

# # Assuming your columns have names
# columns = ['ts_on_min', 'ts_on_max', 'ts_on_mode', 'ts_on_mean',
#            'ts_off_min', 'ts_off_max', 'ts_off_mode', 'ts_off_mean',
#            'tspc_count', 'OCEAN']

# # Create DataFrame
# df = pd.DataFrame(your_data, columns=columns)

# # Scale the data (very important for autoencoders)
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(all_parameters_input)
```

```python
from keras.layers import Input, Dense
from keras.models import Model
def create_autoencoder(input_dim):
    input_layer = Input(shape=(input_dim,))

    # Encoder
    encoded = Dense(32, activation='relu')(input_layer)
    encoded = Dense(16, activation='relu')(encoded)
    encoded = Dense(8, activation='relu')(encoded)

    # Decoder
    decoded = Dense(16, activation='relu')(encoded)
    decoded = Dense(32, activation='relu')(decoded)
    decoded = Dense(input_dim, activation='sigmoid')(decoded)

    return Model(input_layer, decoded)
```

```python
# Create and compile model
input_dim = scaled_data.shape[1]  # number of features
autoencoder = create_autoencoder(input_dim)
autoencoder.compile(optimizer='adam', loss='mse')

# Split data
from sklearn.model_selection import train_test_split
# X_train, X_test = train_test_split(scaled_data, test_size=0.2, random_state=42)
data_length = len(scaled_data)
train_size = int(0.8 * data_length)

X_train = scaled_data[:train_size]
X_test = scaled_data[train_size:]

# Train
history_all = autoencoder.fit(X_train, X_train,
                         epochs=100,
                         batch_size=32,
                         validation_split=0.1,
                         verbose=1)
```

```python
# Get reconstruction error
reconstructed_all = autoencoder.predict(scaled_data)
mse_all = np.mean(np.power(scaled_data - reconstructed_all, 2), axis=1)

# Set threshold
threshold = np.mean(mse_all) + 2 * np.std(mse_all)

# Find anomalies
anomalies = mse_all > threshold
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Z-score normalization function
def calculate_z_scores(mse, threshold):
    mu = np.mean(mse)
    sigma = np.std(mse)
    z_scores = (mse - mu) / sigma
    anomalies = np.where(np.abs(z_scores) > threshold)[0]
    return z_scores, anomalies

# Feature sets and thresholds
feature_sets = {
    'All': {'mse': mse_all, 'threshold': t_all},
    'Device': {'mse': mse_device, 'threshold': t_device},
    'Logon/Logoff': {'mse': mse_log, 'threshold': t_log},
    'PC': {'mse': mse_pc, 'threshold': t_pc}
}

# Create the plot
plt.figure(figsize=(15, 8))

for i, (feature_name, data) in enumerate(feature_sets.items()):
    mse = data['mse']
    threshold = data['threshold']

    # Calculate Z-scores and anomalies
    z_scores, anomalies = calculate_z_scores(mse, threshold)

    # Plot Z-scores for the feature set
    plt.scatter(np.arange(len(z_scores)) + (i * 0.2), z_scores, label=f'Normal ({feature_name})', alpha=0.6)
    plt.scatter(np.arange(len(z_scores[anomalies])) + (i * 0.2), z_scores[anomalies],
                label=f'Anomalies ({feature_name})', color='red', alpha=0.6)

    # Add threshold lines
    plt.axhline(y=threshold, color='r', linestyle='--', label=f'Upper Threshold ({feature_name})')
    plt.axhline(y=-threshold, color='r', linestyle='--', label=f'Lower Threshold ({feature_name})')

    # Print the number of anomalies
    print(f"Number of anomalies for {feature_name}: {len(anomalies)}")

# Customize plot
plt.title('Z-Scores of Reconstruction Errors for Different Feature Sets', size=18)
plt.xlabel('Data Point Index', size=15)
plt.ylabel('Z-Score', size=15)
plt.legend(fontsize=10, loc='upper right')
plt.grid(True, alpha=0.3)

# Show plot
plt.show()
```
